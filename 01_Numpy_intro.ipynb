## What is an Array 

* An array is a data structure used to store multiple values of the same or related type in a single variable, where each value can be accessed using an index.


* In Python, an array is a collection of multiple values stored under one variable name, where each value can be accessed using an index.

```python
import numpy as np 
```

```python
a = np.arange(10)
print(a)
```

```python
a2 = a [np.newaxis, :]
print(a2)
a2.shape
```

```python
a2.shape
```

```python
# 3D Array
a3 = a2[np.newaxis, :]
print (a3)

a3.shape
```
