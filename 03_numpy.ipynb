## Numpy Indexing and Slicing 

```python
import numpy as np 
```

```python
# Create an array 

a = np.array([1,2,3,4,5,6,7,8])
a
```

```python
b = np.arange(0,12,2)
b
```

```python
b[2]
```

```python
b.reshape(2,3)
```

```python
c = b.reshape(2,3)
c
```

```python
c[0]
```

```python
c[(c < 5) & ( c > 7)]
```
